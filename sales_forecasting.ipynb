```python
import pandas as pd

train = pd.read_csv("store-sales-time-series-forecasting/train.csv")
stores = pd.read_csv("store-sales-time-series-forecasting/stores.csv")
```

```python
train['date'] = pd.to_datetime(train['date'])
```

```python
merged_df = train.merge(stores, on = 'store_nbr', how = 'left')

merged_df['month'] = merged_df.date.dt.to_period('M')
```

```python
print(merged_df.groupby('month').sales.sum().sort_values(ascending=False).head(15))
print(merged_df.groupby('store_nbr').sales.sum().sort_values(ascending=False).head(15))
print(merged_df.groupby(['month', 'store_nbr']).sales.sum().sort_values(ascending = False).head(15))
```
